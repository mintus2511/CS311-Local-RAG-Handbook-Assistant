### Baseline Semantic Retrieval Evaluation

The baseline retrieval system was evaluated using three representative questions covering student enrollment, research activities, and student organizations at Fulbright University Vietnam.

The retrieval pipeline used `nomic-embed-text` for query embeddings, ChromaDB for vector similarity search, and a Top-K setting of 5.

All three queries retrieved a top-ranked document chunk containing information directly relevant to the question. The Top-1 vector distances were 0.3018, 0.4284, and 0.3561, respectively.

The retrieval times were 0.083, 0.062, and 0.058 seconds, corresponding to an average retrieval latency of approximately **0.068 seconds per query**.

The results suggest that the baseline retrieval pipeline can successfully identify relevant information from the small, domain-specific knowledge base.

However, lower-ranked results were not uniformly relevant. For example, the enrollment question retrieved some chunks concerning institutional history and research activities rather than enrollment figures. This illustrates a potential limitation of semantic retrieval, where topical similarity does not always guarantee answer relevance.

The initial results are encouraging, but the evaluation is limited to three manually inspected questions. A larger benchmark with ground-truth evidence annotations is required to quantify retrieval performance using metrics such as Recall@K and Mean Reciprocal Rank (MRR).

```python
from langchain_ollama import OllamaLLM
import time

LLM_MODEL = "qwen2.5:0.5b"
TEMPERATURE = 0.7

llm = OllamaLLM(
    model=LLM_MODEL,
    base_url="http://127.0.0.1:11434",
    temperature=TEMPERATURE
)

print("Testing Qwen language model...")

start = time.perf_counter()

response = llm.invoke(
    "What is Retrieval-Augmented Generation? "
    "Answer in one sentence."
)

elapsed = time.perf_counter() - start

print("\nModel response:")
print(response)
print(f"\nInference time: {elapsed:.2f} seconds")
print("[SUCCESS] LLM inference completed!")
```

```python

from langchain_ollama import OllamaLLM
import time

TOP_K = 5
SCORE_THRESHOLD = 0.3

PROMPT_TEMPLATE = """You are a helpful AI assistant that answers questions based on provided documents.

CONTEXT FROM DOCUMENTS:
{context}

---

QUESTION: {question}

INSTRUCTIONS:
1. If the context contains relevant information, answer based on that.
2. If the context is insufficient, use your general knowledge.
3. Be concise - 2-4 sentences unless more detail is needed.

ANSWER:"""

def ask_rag(question):
    start = time.perf_counter()

    # Step 1: Retrieve chunks
    raw_results = vector_db.similarity_search_with_score(
        question, k=TOP_K
    )
    retrieval_time = time.perf_counter() - start

    # Step 2: Convert distance to baseline score
    sources = []
    for doc, distance in raw_results:
        score = 1 / (1 + distance)

        sources.append({
            "source": doc.metadata.get("source", "Unknown"),
            "score": round(score, 4),
            "distance": float(distance),
            "content": doc.page_content
        })

    avg_score = (
        sum(s["score"] for s in sources) / len(sources)
        if sources else 0
    )

    # Step 3: Baseline fallback
    used_fallback = (
        not sources or avg_score < SCORE_THRESHOLD
    )

    if used_fallback:
        context = (
            "No sufficiently relevant documents found. "
            "Use your general knowledge."
        )
    else:
        context = "\n\n---\n\n".join(
            f"[{s['source']}]\n{s['content']}"
            for s in sources
        )

    # Step 4: Construct prompt
    prompt = PROMPT_TEMPLATE.format(
        context=context,
        question=question
    )

    # Step 5: Generate answer
    generation_start = time.perf_counter()
    answer = llm.invoke(prompt).strip()
    generation_time = time.perf_counter() - generation_start

    return {
        "question": question,
        "answer": answer,
        "sources": sources,
        "avg_score": avg_score,
        "used_fallback": used_fallback,
        "retrieval_time": retrieval_time,
        "generation_time": generation_time,
        "total_time": time.perf_counter() - start
    }

print("[SUCCESS] Baseline RAG function initialized!")
```

```python

test_questions = [
    "How many students were enrolled at Fulbright University Vietnam?",
    "What are the main research areas at Fulbright?",
    "What student organizations and activities are available?"
]

rag_results = []

for i, question in enumerate(test_questions, 1):
    print("=" * 65)
    print(f"TEST QUESTION {i}")
    print("=" * 65)

    result = ask_rag(question)
    rag_results.append(result)

    print(f"\nQuestion: {question}")
    print(f"\nAnswer:\n{result['answer']}")

    print("\nTop-5 Sources:")
    for rank, source in enumerate(result["sources"], 1):
        print(
            f"{rank}. {source['source']} "
            f"(score={source['score']:.4f})"
        )

    print(f"\nAverage score: {result['avg_score']:.4f}")
    print(f"Used fallback: {result['used_fallback']}")
    print(f"Retrieval time: {result['retrieval_time']:.3f}s")
    print(f"Generation time: {result['generation_time']:.3f}s")
    print(f"Total time: {result['total_time']:.3f}s")
    print()
```

### Baseline RAG Evaluation and Discussion

#### Experimental Setup

The baseline Retrieval-Augmented Generation (RAG) system was implemented using ChromaDB for vector storage, `nomic-embed-text` for semantic embeddings, and `qwen2.5:0.5b` for answer generation through Ollama.

The system used a chunk size of 1,000 characters, a chunk overlap of 200 characters, a Top-K retrieval setting of 5, a temperature of 0.7, and a baseline relevance-score threshold of 0.3.

Three representative questions were used to evaluate retrieval performance, answer correctness, and inference efficiency.

#### Results

| Metric | Question 1 | Question 2 | Question 3 |
|---|---:|---:|---:|
| Average retrieval score | 0.7191 | 0.6750 | 0.6373 |
| Retrieval time (s) | 0.139 | 0.164 | 0.191 |
| Generation time (s) | 41.652 | 53.058 | 55.315 |
| Total time (s) | 41.792 | 53.222 | 55.506 |
| Fallback activated | No | No | No |

The average retrieval latency was approximately **0.165 seconds**, whereas average generation latency was approximately **50.008 seconds**. The overall average response time was approximately **50.173 seconds**.

#### Observations

**1. Retrieval effectiveness**

For all three questions, the highest-ranked retrieved chunk contained information relevant to the user's query. This indicates that the embedding model and vector database were capable of identifying useful evidence within the small document collection.

However, some lower-ranked chunks were only broadly related to the question and did not necessarily contain the information required for an accurate answer.

**2. Answer correctness and hallucination**

Although the generated answers were generally relevant, several factual inconsistencies were observed.

For the enrollment question, the model correctly identified 832 students in Academic Year 2024–2025 but incorrectly associated the undergraduate and graduate enrollment figures with the university's first undergraduate cohort in 2018.

For the research question, the model identified the main research fields but repeated renewable energy and introduced additional descriptions that were not clearly supported by the retrieved evidence.

For the student organizations question, the model correctly identified several extracurricular activities but incorrectly associated more than 1,500 student registrations with startup development rather than technology-related workshops and events.

These findings demonstrate that retrieving relevant documents does not automatically guarantee faithful answer generation.

**3. Computational efficiency**

Retrieval was considerably faster than answer generation. While retrieving the five most similar chunks required less than 0.2 seconds per question, generating answers required approximately 42–55 seconds on the CPU runtime.

The results suggest that language-model inference is the dominant computational bottleneck in the tested configuration. Further profiling would be required to distinguish the effects of model execution, prompt length, and output length.

**4. Limitations of the baseline design**

The baseline system uses an average transformed distance score to determine whether retrieved evidence is sufficiently relevant. However, this score is not a calibrated probability of answer correctness, and averaging the Top-K scores may hide differences in relevance between individual chunks.

Additionally, the prompt permits the model to use general knowledge when retrieved context is insufficient. While this can improve answer coverage, it may also introduce unsupported information into responses intended to be grounded in the provided documents.

```python

import pymupdf
from pathlib import Path
from langchain_core.documents import Document

PDF_DIR = PROJECT_ROOT / "data" / "pdf"
PDF_PATH = PDF_DIR / "CS_MajorHandbook.pdf"

assert PDF_PATH.exists(), f"PDF not found: {PDF_PATH}"

def load_pdf_documents(pdf_path):
    documents = []
    skipped_pages = []

    with pymupdf.open(str(pdf_path)) as pdf:
        for page_index, page in enumerate(pdf):
            text = page.get_text("text").strip()

            if not text:
                skipped_pages.append(page_index + 1)
                continue

            doc = Document(
                page_content=text,
                metadata={
                    "source": pdf_path.name,
                    "page": page_index + 1,
                    "file_type": "pdf"
                }
            )

            documents.append(doc)

    return documents, skipped_pages


pdf_documents, skipped_pages = load_pdf_documents(PDF_PATH)

print("=" * 60)
print("Task 1 - PDF Document Loading")
print("=" * 60)

print("PDF file:", PDF_PATH.name)
print("Documents loaded:", len(pdf_documents))
print("Pages skipped:", skipped_pages)

print("\nSample document metadata:")
print(pdf_documents[0].metadata)

print("\nSample content:")
print(pdf_documents[0].page_content[:500])

assert len(pdf_documents) == 18
assert skipped_pages == [10]

print("\n[SUCCESS] PDF documents loaded!")
```

```python

from langchain_ollama import OllamaLLM
import time

pdf_llm = OllamaLLM(
    model="qwen2.5:0.5b",
    base_url="http://127.0.0.1:11434",
    temperature=0.0,
    num_predict=250
)

PDF_RAG_PROMPT = """You are an academic assistant answering questions
about Fulbright University Vietnam's Computer Science handbook.

Use ONLY the provided document excerpts.

RULES:
1. Answer based only on the context.
2. Do not invent facts, numbers, or course requirements.
3. If the information is insufficient, say:
   "I could not find sufficient information in the handbook."
4. Be concise and avoid repeating information.
5. Cite the supporting PDF page number.
6. Do not confuse major and minor requirements.

DOCUMENT EXCERPTS:
{context}

QUESTION:
{question}

ANSWER:
"""

def ask_pdf_rag(question, top_k=5):
    start = time.perf_counter()

    results = pdf_vector_db.similarity_search_with_score(
        question, k=top_k
    )

    retrieval_time = time.perf_counter() - start

    context_parts = []

    sources = []

    for rank, (doc, distance) in enumerate(results, 1):
        page = doc.metadata.get("page")
        source = doc.metadata.get("source")

        context_parts.append(
            f"[Source {rank}: {source}, PDF page {page}]\n"
            f"{doc.page_content}"
        )

        sources.append({
            "source": source,
            "page": page,
            "distance": float(distance),
            "content": doc.page_content
        })

    context = "\n\n---\n\n".join(context_parts)

    prompt = PDF_RAG_PROMPT.format(
        context=context,
        question=question
    )

    generation_start = time.perf_counter()
    answer = pdf_llm.invoke(prompt).strip()
    generation_time = time.perf_counter() - generation_start

    return {
        "question": question,
        "answer": answer,
        "sources": sources,
        "retrieval_time": retrieval_time,
        "generation_time": generation_time,
        "total_time": time.perf_counter() - start
    }

print("[SUCCESS] PDF RAG initialized!")
```

```python

import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Build keyword index from existing PDF chunks
tfidf_vectorizer = TfidfVectorizer(
    lowercase=True,
    ngram_range=(1, 2),
    sublinear_tf=True
)

chunk_texts = [doc.page_content for doc in pdf_chunks]
tfidf_matrix = tfidf_vectorizer.fit_transform(chunk_texts)

# Prepare mapping between Chroma IDs and chunk positions
chunk_id_to_index = {
    f"cs_pdf_chunk_{i:04d}": i
    for i in range(len(pdf_chunks))
}

def hybrid_retrieve(question, top_k=5, candidate_k=15, rrf_k=60):

    # 1. Semantic retrieval
    semantic_results = pdf_vector_db.similarity_search_with_score(
        question, k=min(candidate_k, len(pdf_chunks))
    )

    semantic_ranks = {}
    for rank, (doc, distance) in enumerate(semantic_results, 1):
        # Match by source, page and content
        for idx, candidate in enumerate(pdf_chunks):
            if (
                candidate.metadata == doc.metadata
                and candidate.page_content == doc.page_content
            ):
                semantic_ranks[idx] = rank
                break

    # 2. Keyword retrieval
    query_vector = tfidf_vectorizer.transform([question])
    lexical_scores = cosine_similarity(
        query_vector, tfidf_matrix
    ).ravel()

    lexical_order = np.argsort(-lexical_scores)
    lexical_ranks = {
        int(idx): rank
        for rank, idx in enumerate(
            lexical_order[:candidate_k], 1
        )
    }

    # 3. Reciprocal Rank Fusion
    fused_scores = {}

    for idx in set(semantic_ranks) | set(lexical_ranks):
        score = 0.0

        if idx in semantic_ranks:
            score += 1 / (rrf_k + semantic_ranks[idx])

        if idx in lexical_ranks:
            score += 1 / (rrf_k + lexical_ranks[idx])

        fused_scores[idx] = score

    ranked_indices = sorted(
        fused_scores,
        key=lambda idx: fused_scores[idx],
        reverse=True
    )[:top_k]

    return [
        {
            "document": pdf_chunks[idx],
            "fusion_score": fused_scores[idx],
            "semantic_rank": semantic_ranks.get(idx),
            "keyword_rank": lexical_ranks.get(idx)
        }
        for idx in ranked_indices
    ]

print("[SUCCESS] Hybrid retriever initialized!")
```

### Root Cause Analysis: Document Structure and Chunk Boundaries

An inspection of the 58 PDF chunks revealed that important academic section headings were frequently separated from their corresponding content.

For example, the heading "Major Learning Outcomes" appeared in chunk 12, while the actual MLO definitions were primarily located in chunk 13. Similarly, the major and minor requirements on page 7 were split across multiple chunks, with some chunks containing information from both sections.

The document table of contents also introduced repeated academic keywords that could attract retrieval results without providing the actual answers.

These findings identify document segmentation as a plausible contributor to the observed retrieval errors. Standard character-based chunking preserves approximate chunk sizes but does not necessarily preserve logical document sections.

A structure-aware preprocessing strategy will therefore be investigated to improve contextual coherence and distinguish closely related academic requirements.

```python

import re
from collections import Counter
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Patterns identify headings, not answer values.
SECTION_PATTERN = re.compile(
    r"(?im)^[ \t]*("
    r"Major Requirements\s*\(\d+\s+credits\)"
    r"|Minor Requirements\s*\(\d+\s+credits\)"
    r"|Program Learning Outcomes(?:\s*\(PLOs\))?"
    r"|Short Name\s*\nMajor LOs"
    r"|Course Alignment with the Program Learning Outcomes(?:\s*\(PLOs\))?"
    r")[ \t]*$"
)

structure_splitter = RecursiveCharacterTextSplitter(
    chunk_size=850,
    chunk_overlap=150
)

def split_into_sections(text):
    matches = list(SECTION_PATTERN.finditer(text))

    if not matches:
        return [("General content", text)]

    sections = []

    # Content preceding the first heading
    if text[:matches[0].start()].strip():
        sections.append((
            "General content",
            text[:matches[0].start()].strip()
        ))

    for i, match in enumerate(matches):
        start = match.end()
        end = (
            matches[i + 1].start()
            if i + 1 < len(matches)
            else len(text)
        )

        heading = " ".join(match.group(1).split())
        content = text[start:end].strip()

        if content:
            sections.append((heading, content))

    return sections


structured_chunks = []

for page_doc in pdf_documents:
    page = page_doc.metadata["page"]
    text = page_doc.page_content

    # Skip the table of contents, identified by repeated dot leaders.
    dot_leader_lines = len(re.findall(r"\.{5,}", text))
    if dot_leader_lines >= 5:
        continue

    for section, section_text in split_into_sections(text):
        # Keep section headings attached to every child chunk.
        prefix = f"Section: {section}\n"

        child_docs = structure_splitter.split_documents([
            Document(
                page_content=section_text,
                metadata={
                    "source": page_doc.metadata["source"],
                    "page": page,
                    "file_type": "pdf",
                    "section": section
                }
            )
        ])

        for child in child_docs:
            structured_chunks.append(Document(
                page_content=prefix + child.page_content,
                metadata=child.metadata
            ))

print("=" * 65)
print("Structure-Aware Chunking Results")
print("=" * 65)

print("Original chunks:", len(pdf_chunks))
print("Structure-aware chunks:", len(structured_chunks))

lengths = [len(doc.page_content) for doc in structured_chunks]

print("Average chunk length:", round(sum(lengths) / len(lengths), 2))
print("Maximum chunk length:", max(lengths))
print("Minimum chunk length:", min(lengths))

section_counts = Counter(
    doc.metadata["section"] for doc in structured_chunks
)

print("\nSection distribution:")
for section, count in section_counts.items():
    print(f"{section}: {count}")

print("\nSample section-aware chunks from page 7:")

for doc in structured_chunks:
    if doc.metadata["page"] == 7:
        print("\nSection:", doc.metadata["section"])
        print(doc.page_content[:250])
```

### Improvement Experiment 5: Query Expansion and Reciprocal Rank Fusion

A query expansion experiment was conducted to improve retrieval coverage for questions about the Computer Science program's learning outcomes.

Three queries were used: the original question, a variant referring explicitly to Program Learning Outcomes (PLOs), and another referring to Major Learning Outcomes (MLOs).

The expanded retrieval process successfully identified the formal Major Learning Outcomes section on PDF page 5. This section ranked first for the MLO-specific query, demonstrating the importance of terminology alignment between the question and source document.

However, after Reciprocal Rank Fusion, the MLO section appeared at Rank 6, outside the Top-5 context window used by the baseline RAG system.

Retrieval latency was 0.561 seconds for the three-query experiment, compared with 0.177 seconds for the earlier single-query structure-aware retrieval test. This difference is descriptive rather than a controlled performance benchmark.

**Discussion:** Query expansion improved the diversity of relevant evidence discovered, but standard RRF did not guarantee that complementary information was retained within the final context window.

These findings motivate testing an evidence-selection strategy that preserves relevant results from multiple query variants rather than relying exclusively on their combined ranking scores.

```python

import time

def balanced_multi_query_retrieve(
    queries,
    top_k=5,
    per_query_k=10
):
    retrieval_lists = []

    # Retrieve candidates for each query variant
    for query in queries:
        results = structured_db.similarity_search_with_score(
            query, k=per_query_k
        )
        retrieval_lists.append(results)

    selected = []
    seen = set()

    # Round-robin selection across query variants
    for rank_index in range(per_query_k):
        for query_index, results in enumerate(retrieval_lists):

            if rank_index >= len(results):
                continue

            doc, distance = results[rank_index]

            key = (
                doc.metadata.get("source"),
                doc.metadata.get("page"),
                doc.page_content
            )

            if key in seen:
                continue

            seen.add(key)

            selected.append({
                "document": doc,
                "distance": float(distance),
                "query_index": query_index,
                "original_rank": rank_index + 1
            })

            if len(selected) >= top_k:
                return selected

    return selected


start = time.perf_counter()

balanced_results = balanced_multi_query_retrieve(
    expanded_queries,
    top_k=5,
    per_query_k=10
)

balanced_time = time.perf_counter() - start

print("=" * 65)
print("Balanced Multi-Query Retrieval")
print("=" * 65)

print(f"Retrieval time: {balanced_time:.3f}s")

for rank, result in enumerate(balanced_results, 1):
    doc = result["document"]

    print(f"\nRank {rank}")
    print(f"PDF Page: {doc.metadata['page']}")
    print(f"Section: {doc.metadata['section']}")
    print(f"Query variant: {result['query_index']}")
    print(f"Original query rank: {result['original_rank']}")
    print(f"Distance: {result['distance']:.4f}")
    print("Content:")
    print(doc.page_content[:350])

mlo_found = any(
    "MLO 1" in item["document"].page_content
    and "MLO 2" in item["document"].page_content
    for item in balanced_results
)

print("\nMLO evidence included:", mlo_found)
```

### Quantitative Retrieval Evaluation Methodology

A page-level retrieval evaluation was designed to compare the original semantic retrieval pipeline with the structure-aware retrieval pipeline.

The evaluation uses 18 answerable questions from the annotated Computer Science Major Handbook benchmark. Two questions without supporting evidence pages are excluded from retrieval ranking metrics and reserved for a separate abstention evaluation.

For each question, the retrievers return the five highest-ranked chunks. Retrieved PDF page numbers are compared with the annotated supporting pages.

The evaluation measures Hit@1, Hit@3, Hit@5, Mean Reciprocal Rank at 5 (MRR@5), All-Evidence@5, and retrieval latency.

An evidence hit is counted when at least one retrieved chunk originates from an annotated supporting page. Consequently, these metrics measure page-level evidence retrieval rather than verified chunk-level relevance.

The evaluation additionally separates previously used development questions from newer evaluation candidates to reduce overinterpretation of results on questions that influenced system design.

Both retrieval systems are evaluated using the same question set and Top-K setting, without requiring language-model generation.

```python

import pandas as pd
import numpy as np
import time
from pathlib import Path

# Evaluate only questions with supporting evidence.
eval_df = benchmark_df[
    benchmark_df["answerable"] == True
].copy()

def parse_evidence_pages(value):
    if pd.isna(value) or not str(value).strip():
        return set()

    return {
        int(page.strip())
        for page in str(value).split(",")
    }


def evaluate_retriever(
    vector_db,
    questions_df,
    system_name,
    max_k=5
):
    records = []

    for _, row in questions_df.iterrows():
        question = row["question"]
        evidence_pages = parse_evidence_pages(
            row["evidence_pages"]
        )

        start = time.perf_counter()

        retrieved = vector_db.similarity_search_with_score(
            question,
            k=max_k
        )

        retrieval_time = time.perf_counter() - start

        retrieved_pages = [
            int(doc.metadata["page"])
            for doc, distance in retrieved
        ]

        first_relevant_rank = next(
            (
                rank
                for rank, page in enumerate(retrieved_pages, 1)
                if page in evidence_pages
            ),
            None
        )

        record = {
            "system": system_name,
            "id": row["id"],
            "split": row["split"],
            "question": question,
            "evidence_pages": sorted(evidence_pages),
            "retrieved_pages": retrieved_pages,
            "first_relevant_rank": first_relevant_rank,
            "retrieval_time": retrieval_time
        }

        for k in [1, 3, 5]:
            top_pages = set(retrieved_pages[:k])

            record[f"hit@{k}"] = int(
                len(top_pages & evidence_pages) > 0
            )

            record[f"all_evidence@{k}"] = int(
                evidence_pages.issubset(top_pages)
            )

        record["rr@5"] = (
            1 / first_relevant_rank
            if first_relevant_rank is not None
            else 0.0
        )

        records.append(record)

    return pd.DataFrame(records)


# Run both retrieval systems on identical questions.
baseline_eval = evaluate_retriever(
    vector_db=pdf_vector_db,
    questions_df=eval_df,
    system_name="Original Semantic"
)

structured_eval = evaluate_retriever(
    vector_db=structured_db,
    questions_df=eval_df,
    system_name="Structure-Aware"
)

retrieval_eval_df = pd.concat(
    [baseline_eval, structured_eval],
    ignore_index=True
)

# Summarize aggregate metrics.
metrics = [
    "hit@1",
    "hit@3",
    "hit@5",
    "all_evidence@5",
    "rr@5",
    "retrieval_time"
]

summary = (
    retrieval_eval_df
    .groupby("system")[metrics]
    .mean()
    .round(4)
    .rename(columns={
        "rr@5": "MRR@5",
        "retrieval_time": "Avg latency (s)"
    })
)

print("=" * 70)
print("RETRIEVAL BENCHMARK — 18 ANSWERABLE QUESTIONS")
print("=" * 70)

print("\nAggregate results:")
print(summary.to_string())

print("\nPer-question comparison:")
comparison = (
    retrieval_eval_df
    .pivot(
        index="id",
        columns="system",
        values=["hit@1", "hit@5", "rr@5"]
    )
)
print(comparison.to_string())

# Save results to Google Drive.
RESULTS_DIR = PROJECT_ROOT / "results"
RESULTS_DIR.mkdir(parents=True, exist_ok=True)

retrieval_eval_df.to_csv(
    RESULTS_DIR / "retrieval_evaluation_detailed.csv",
    index=False
)

summary.to_csv(
    RESULTS_DIR / "retrieval_evaluation_summary.csv"
)

print("\n[SUCCESS] Retrieval benchmark completed!")
print("Detailed:", RESULTS_DIR / "retrieval_evaluation_detailed.csv")
print("Summary:", RESULTS_DIR / "retrieval_evaluation_summary.csv")
```

### Course Catalogue Structure Inspection

A document structure inspection was conducted to investigate retrieval failures involving course prerequisites.

The original PDF text showed a consistent pattern for many catalogue entries, consisting of a course code and title followed by prerequisites and a course description.

For example, the CS201 entry on PDF page 14 contained the heading "CS201: Data Structures & Algorithms" immediately followed by "Prerequisites: CS101".

However, the previous character-based chunking pipelines separated the course heading from its prerequisite information, making the retrieved evidence ambiguous.

The inspection detected 20 course headings across pages 13–18. No course heading matching the tested pattern was detected on page 19.

These findings motivate a course-aware segmentation approach that treats each course entry as a logical document unit while preserving source page information. This approach may improve retrieval for questions involving prerequisites, course descriptions, and individual course requirements.

```python

import re
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Only use the course-catalogue portion inspected in Cell 53.
catalogue_pages = sorted(
    [
        doc for doc in pdf_documents
        if 13 <= doc.metadata["page"] <= 19
    ],
    key=lambda doc: doc.metadata["page"]
)

# Combine pages, preserving character-offset-to-page mapping.
catalogue_text = ""
page_spans = []

for doc in catalogue_pages:
    start = len(catalogue_text)
    catalogue_text += doc.page_content + "\n\n"
    end = len(catalogue_text)
    page_spans.append((start, end, doc.metadata["page"]))

course_pattern = re.compile(
    r"(?im)^[ \t]*(CS\s*\d{3}(?:/\d{3})?)\s*:\s*([^\n]+)"
)

course_matches = list(course_pattern.finditer(catalogue_text))

def pages_for_span(start, end):
    return [
        page for page_start, page_end, page in page_spans
        if page_start < end and page_end > start
    ]

course_documents = []

for i, match in enumerate(course_matches):
    start = match.start()
    end = (
        course_matches[i + 1].start()
        if i + 1 < len(course_matches)
        else len(catalogue_text)
    )

    course_code = re.sub(
        r"\s+", "", match.group(1)
    ).upper()

    course_title = match.group(2).strip()

    course_body = catalogue_text[match.end():end].strip()
    pages = pages_for_span(start, end)

    course_documents.append(
        Document(
            page_content=(
                f"Course: {course_code}: {course_title}\n"
                f"{course_body}"
            ),
            metadata={
                "source": "CS_MajorHandbook.pdf",
                "file_type": "pdf",
                "page": pages[0],
                "end_page": pages[-1],
                "course_code": course_code,
                "course_title": course_title,
                "section": "Course Catalogue"
            }
        )
    )

print("=" * 65)
print("COURSE-AWARE EXTRACTION")
print("=" * 65)

print("Course entries detected:", len(course_documents))

for doc in course_documents:
    print(
        f"{doc.metadata['course_code']} | "
        f"Page {doc.metadata['page']}-"
        f"{doc.metadata['end_page']} | "
        f"{doc.metadata['course_title']}"
    )

cs201_docs = [
    doc for doc in course_documents
    if doc.metadata["course_code"] == "CS201"
]

assert len(cs201_docs) == 1, "Expected exactly one CS201 entry."

print("\n" + "=" * 65)
print("CS201 COURSE DOCUMENT PREVIEW")
print("=" * 65)

print(cs201_docs[0].page_content[:1200])

assert "Prerequisites: CS101" in cs201_docs[0].page_content

print("\n[SUCCESS] Course entries extracted with metadata!")
```

```python

course_splitter = RecursiveCharacterTextSplitter(
    chunk_size=850,
    chunk_overlap=150
)

course_chunks = []

for course_doc in course_documents:
    metadata = course_doc.metadata.copy()

    prefix = (
        f"Course: {metadata['course_code']}: "
        f"{metadata['course_title']}\n"
    )

    # Remove the heading, which we will repeat on every chunk.
    body = course_doc.page_content[len(prefix):]

    # Repeating the heading uses part of the chunk budget.
    available_size = 850 - len(prefix)

    local_splitter = RecursiveCharacterTextSplitter(
        chunk_size=available_size,
        chunk_overlap=min(150, available_size // 4)
    )

    pieces = local_splitter.split_text(body)

    for piece_index, piece in enumerate(pieces):
        course_chunks.append(
            Document(
                page_content=prefix + piece,
                metadata={
                    **metadata,
                    "piece_index": piece_index
                }
            )
        )

print("=" * 65)
print("COURSE-AWARE CHUNKING")
print("=" * 65)

print("Course entries:", len(course_documents))
print("Retrieval chunks:", len(course_chunks))

lengths = [len(doc.page_content) for doc in course_chunks]

print("Average chunk length:", round(sum(lengths)/len(lengths), 2))
print("Maximum chunk length:", max(lengths))

print("\nCS201 chunks:")

for doc in course_chunks:
    if doc.metadata["course_code"] == "CS201":
        print("\nMetadata:", doc.metadata)
        print(doc.page_content[:650])
        print("-" * 50)

assert max(lengths) <= 850

print("\n[SUCCESS] Course-aware chunking completed!")
```

A course-aware vector database containing 34 chunks was evaluated using the question "What prerequisites are listed for CS201?"

This demonstrates a distinction between two relationships: the prerequisite of CS201 and courses that require CS201.

The embedding-based retriever identified lexical and semantic similarities but failed to distinguish the role of the course code in the question.

All three evaluated retrieval pipelines failed to return self-contained evidence for the target course within their Top-5 results.

These findings motivate metadata-filtered retrieval, which explicitly identifies the target course code and restricts semantic search to the corresponding course documents.

This approach combines symbolic entity matching with semantic retrieval, potentially improving precision for course-specific questions without requiring a larger embedding model.

```python

import re
import time

def extract_course_codes(question):
    """
    Extract course codes such as CS201, CS 201, CS-201.
    """
    matches = re.findall(
        r"\bCS[\s-]*(\d{3})\b",
        question,
        flags=re.IGNORECASE
    )

    return list(dict.fromkeys(
        f"CS{number}" for number in matches
    ))


def retrieve_target_course(question, top_k=5):
    """
    Retrieves documents for an explicitly referenced
    target course.

    Intended for questions about the named course itself,
    not inverse questions such as:
    'Which courses require CS201?'
    """
    course_codes = extract_course_codes(question)

    if len(course_codes) != 1:
        return {
            "status": "unsupported_course_selection",
            "course_codes": course_codes,
            "results": []
        }

    target_code = course_codes[0]

    start = time.perf_counter()

    results = course_db.similarity_search_with_score(
        question,
        k=top_k,
        filter={"course_code": target_code}
    )

    elapsed = time.perf_counter() - start

    return {
        "status": "success" if results else "not_found",
        "target_course": target_code,
        "results": results,
        "retrieval_time": elapsed
    }


question = "What prerequisites are listed for CS 201?"

result = retrieve_target_course(question)

print("=" * 65)
print("COURSE-CODE-AWARE RETRIEVAL")
print("=" * 65)

print("Question:", question)
print("Detected course:", result.get("target_course"))
print("Status:", result["status"])

if result["status"] == "success":
    print(f"Retrieval time: {result['retrieval_time']:.3f}s")

    for rank, (doc, distance) in enumerate(
        result["results"], 1
    ):
        print(f"\nRank {rank}")
        print("Course:", doc.metadata["course_code"])
        print("Title:", doc.metadata["course_title"])
        print(
            "PDF pages:",
            doc.metadata["page"],
            "-",
            doc.metadata["end_page"]
        )
        print(f"Distance: {distance:.4f}")
        print("\nContent:")
        print(doc.page_content[:600])

    first_doc = result["results"][0][0]

    assert first_doc.metadata["course_code"] == "CS201"
    assert "Prerequisites: CS101" in first_doc.page_content

    print("\n[SUCCESS] Correct course evidence retrieved!")
else:
    print("No matching target course was retrieved.")
```

### Improvement Experiment 8: Course-Code-Aware Metadata Filtering

A metadata-filtered retrieval strategy was implemented to address failures in identifying the target course within prerequisite-related questions.

The method extracts an explicitly mentioned course code from the user question and applies a ChromaDB metadata filter before performing semantic retrieval.

For the CS201 prerequisite question, the filtered retriever returned a self-contained chunk containing both the course title and its prerequisite, CS101, at Rank 1. Retrieval latency was 0.228 seconds.

Additional tests confirmed that the method correctly identified and retrieved documents for CS311, CS302, and CS203.

However, the CS302 and CS203 tests also revealed that semantic search may return a later chunk from the correct course entry rather than the introductory chunk containing the most important course information.

**Discussion:** Metadata filtering successfully resolves ambiguity between courses that require CS201 and the prerequisites required to take CS201.

Nevertheless, identifying the correct course does not guarantee selection of the most appropriate section within its description.

This motivates combining entity-based filtering with field-aware selection, particularly for structured attributes such as prerequisites, course titles, and course descriptions.

The current findings establish improvement for the diagnostic CS201 question but do not yet demonstrate overall accuracy on unseen course-related questions.

```python

import re
import time

# Course documents were extracted directly from the handbook.
course_lookup = {
    doc.metadata["course_code"]: doc
    for doc in course_documents
}

def detect_course_intent(question):
    q = question.lower()

    # Inverse relationship: which courses require this course?
    if re.search(
        r"(which|what)\s+courses?.*"
        r"(require|need|depend on)",
        q
    ):
        return "inverse_prerequisite"

    if "prerequisit" in q or "required before" in q:
        return "prerequisite"

    return "course_overview"


def field_aware_course_retrieve(question):
    start = time.perf_counter()

    codes = extract_course_codes(question)
    intent = detect_course_intent(question)

    if len(codes) != 1:
        return {
            "status": "unsupported_course_selection",
            "intent": intent,
            "results": []
        }

    target_code = codes[0]

    # Inverse questions need a separate search strategy.
    if intent == "inverse_prerequisite":
        return {
            "status": "inverse_query_requires_separate_retrieval",
            "course_code": target_code,
            "intent": intent,
            "results": []
        }

    doc = course_lookup.get(target_code)

    if doc is None:
        return {
            "status": "course_not_found",
            "course_code": target_code,
            "intent": intent,
            "results": []
        }

    if intent == "prerequisite":
        # Extract the prerequisite field from the actual PDF text.
        match = re.search(
            r"(?im)^\s*Prerequisites?\s*:\s*([^\n]+)",
            doc.page_content
        )

        if match:
            evidence = (
                f"Course: {target_code}: "
                f"{doc.metadata['course_title']}\n"
                f"Prerequisites: {match.group(1).strip()}"
            )
        else:
            evidence = doc.page_content
    else:
        evidence = doc.page_content

    return {
        "status": "success",
        "course_code": target_code,
        "intent": intent,
        "evidence": evidence,
        "metadata": doc.metadata,
        "retrieval_time": time.perf_counter() - start
    }


print("[SUCCESS] Field-aware course retriever initialized!")
```

### Improvement Experiment 9: Field-Aware Course Information Extraction

A field-aware retrieval method was implemented to improve factual extraction for course-specific questions.

The system first identifies the course code mentioned in the question and determines whether the user is requesting prerequisite information or a general course description.

For prerequisite questions, the system directly extracts the relevant field from the corresponding course document. General course questions retrieve the complete course description.

**Results:** The method correctly extracted CS101 as the prerequisite for CS201, CORE105 (Design & Systems Thinking) for CS203, and CS313 for CS311.

The system also retrieved the relevant course description for CS302. An inverse prerequisite question was correctly identified as requiring a different retrieval strategy rather than being answered using the target course's own prerequisites.

The observed lookup and extraction times were below one millisecond for the successful tests.

**Discussion:** Field-aware retrieval demonstrates the potential benefits of combining structured document parsing with deterministic information extraction. For well-defined academic attributes, this approach avoids some of the factual errors observed during language-model generation.

However, the measured times exclude preprocessing and indexing costs. Furthermore, the current rule-based method only supports recognized question patterns and available course metadata.

The results support a hybrid architecture in which structured academic information is handled through deterministic retrieval, while more complex questions are processed through semantic RAG and language-model generation.

```python

import re
import time

def get_course_prerequisites(course_doc):
    """
    Extract the prerequisite field from a course entry.
    Returns None if the field cannot be found.
    """
    match = re.search(
        r"(?im)^[ \t]*Prerequisites?\s*:\s*([^\n]+)",
        course_doc.page_content
    )

    return match.group(1).strip() if match else None


def find_courses_requiring(target_code):
    """
    Find courses whose prerequisite field explicitly
    references the target course code.
    """
    target_code = target_code.upper().replace(" ", "")
    matches = []

    for doc in course_documents:
        prerequisite_text = get_course_prerequisites(doc)

        if not prerequisite_text:
            continue

        # Normalize course-code formatting.
        normalized = re.sub(
            r"\b([A-Za-z]+)[\s-]+(\d{3})\b",
            r"\1\2",
            prerequisite_text.upper()
        )

        if re.search(
            rf"(?<![A-Z0-9]){re.escape(target_code)}(?![A-Z0-9])",
            normalized
        ):
            matches.append({
                "course_code": doc.metadata["course_code"],
                "course_title": doc.metadata["course_title"],
                "prerequisites": prerequisite_text,
                "page": doc.metadata["page"],
                "end_page": doc.metadata["end_page"]
            })

    return matches


question = "Which courses require CS201?"

start = time.perf_counter()

course_codes = extract_course_codes(question)
intent = detect_course_intent(question)

assert intent == "inverse_prerequisite"
assert course_codes == ["CS201"]

inverse_results = find_courses_requiring("CS201")

elapsed = time.perf_counter() - start

print("=" * 65)
print("INVERSE PREREQUISITE RETRIEVAL")
print("=" * 65)

print("Question:", question)
print("Intent:", intent)
print("Target course:", course_codes[0])

print("\nMatching courses:", len(inverse_results))

for item in inverse_results:
    print(
        f"\n{item['course_code']}: {item['course_title']}"
    )
    print("Prerequisites:", item["prerequisites"])
    print(
        f"Source: CS_MajorHandbook.pdf, "
        f"PDF pages {item['page']}-{item['end_page']}"
    )

print(f"\nLookup time: {elapsed:.4f}s")

print("\n[SUCCESS] Inverse prerequisite search completed!")
```

### Improvement Experiment 10: Inverse Prerequisite Retrieval

An inverse prerequisite retrieval function was developed to identify courses that explicitly list a specified course as a prerequisite.

The method scans the prerequisite fields extracted from course catalogue entries and performs normalized course-code matching without language-model inference.

For the query "Which courses require CS201?", the system identified eight matching courses: CS207, CS210, CS211, CS212, CS302, CS304, CS307, and CS312.

The results preserved additional prerequisite conditions where present. For example, CS302 requires both CS201 and MATH202, while CS312 lists CS201 and MATH103.

The observed lookup time was 0.0025 seconds, excluding document preprocessing and loading.

**Discussion:** Deterministic field extraction is particularly suitable for explicit academic relationships such as direct prerequisites. It avoids confusing a course's own prerequisites with courses that depend on it.

However, the method is limited to successfully extracted catalogue entries and explicit prerequisite fields. It does not currently handle indirect dependencies, alternative prerequisites requiring complex interpretation, or missing course records.

The experiment supports integrating symbolic course lookup with semantic RAG in a unified academic information system.

```python

import re
import time

def unified_academic_chat(question, top_k=5):
    start = time.perf_counter()

    codes = extract_course_codes(question)
    intent = detect_course_intent(question)

    # Route 1: Explicit course-code questions
    if len(codes) == 1:
        code = codes[0]

        # 1A. Which courses require the specified course?
        if intent == "inverse_prerequisite":
            matches = find_courses_requiring(code)

            if matches:
                lines = [
                    f"{item['course_code']}: {item['course_title']} "
                    f"(Prerequisites: {item['prerequisites']}) "
                    f"[PDF pp. {item['page']}-{item['end_page']}]"
                    for item in matches
                ]
                answer = (
                    f"Courses explicitly listing {code} "
                    "as a prerequisite:\n" + "\n".join(lines)
                )
            else:
                answer = (
                    f"No matching direct prerequisites for {code} "
                    "were found in the indexed course catalogue."
                )

            return {
                "question": question,
                "answer": answer,
                "route": "inverse_prerequisite",
                "sources": matches,
                "total_time": time.perf_counter() - start
            }

        # 1B. Questions about the named course
        result = field_aware_course_retrieve(question)

        if result["status"] == "success":
            metadata = result["metadata"]

            if intent == "prerequisite":
                answer = result["evidence"]
            else:
                # Extractive display, not LLM generation.
                answer = result["evidence"]

            return {
                "question": question,
                "answer": answer,
                "route": "course_lookup",
                "sources": [{
                    "source": metadata["source"],
                    "page": metadata["page"],
                    "end_page": metadata["end_page"],
                    "course_code": code
                }],
                "total_time": time.perf_counter() - start
            }

        # Don't invent details for an unknown course.
        return {
            "question": question,
            "answer": (
                f"No verified catalogue entry was found for {code}."
            ),
            "route": "course_not_found",
            "sources": [],
            "total_time": time.perf_counter() - start
        }

    # Multiple named courses are not supported by
    # the deterministic single-course lookup.
    # They use general semantic retrieval instead.

    # Route 2: General academic RAG
    results = structured_db.similarity_search_with_score(
        question,
        k=top_k
    )

    context_parts = []
    sources = []

    for rank, (doc, distance) in enumerate(results, 1):
        source = doc.metadata["source"]
        page = doc.metadata["page"]

        context_parts.append(
            f"[Source {rank}: {source}, PDF page {page}]\n"
            f"{doc.page_content}"
        )

        sources.append({
            "source": source,
            "page": page,
            "section": doc.metadata.get("section"),
            "distance": float(distance)
        })

    context = "\n\n---\n\n".join(context_parts)

    prompt = PDF_RAG_PROMPT.format(
        context=context,
        question=question
    )

    answer = pdf_llm.invoke(prompt).strip()

    return {
        "question": question,
        "answer": answer,
        "route": "general_rag",
        "sources": sources,
        "total_time": time.perf_counter() - start
    }

print("[SUCCESS] Unified academic chatbot initialized!")
```

### Abstention Diagnostics: Answerable and Unanswerable Queries

An abstention diagnostic experiment was conducted using 20 benchmark questions, including 18 questions supported by the indexed handbook text and two questions without supporting evidence.

The structure-aware retriever was used to collect the Top-1 and mean Top-5 vector distances for each question.

The answerable questions had a mean Top-1 distance of 0.4220, ranging from 0.2886 to 0.6162. The two unanswerable questions had Top-1 distances of 0.7385 and 0.8151, with a mean of approximately 0.7768.

No overlap was observed between the two groups in this small benchmark.

Manual inspection showed that the retrieved chunks for the unanswerable questions discussed unrelated program information, such as graduate career paths and educational objectives, rather than the requested facts.

**Discussion:** These results suggest that retrieval distance may provide a useful signal for detecting insufficiently relevant evidence. However, distance is not a calibrated measure of factual correctness or answerability.

The absence of overlap in this dataset does not guarantee reliable separation for other questions or documents. The unanswerable sample consists of only two examples, and some answerable questions may use unfamiliar phrasing that produces relatively high distances.

Therefore, a distance-based abstention mechanism should be evaluated on additional challenging questions before being adopted as a final decision rule.

```python

import time

EXPERIMENTAL_DISTANCE_THRESHOLD = 0.68

def check_evidence_sufficiency(question, threshold=0.68, top_k=5):
    start = time.perf_counter()

    results = structured_db.similarity_search_with_score(
        question, k=top_k
    )

    retrieval_time = time.perf_counter() - start

    if not results:
        return {
            "question": question,
            "decision": "abstain",
            "top1_distance": None,
            "sources": [],
            "retrieval_time": retrieval_time
        }

    top1_distance = float(results[0][1])

    decision = (
        "retrieve"
        if top1_distance <= threshold
        else "abstain"
    )

    return {
        "question": question,
        "decision": decision,
        "top1_distance": top1_distance,
        "sources": [
            {
                "page": doc.metadata.get("page"),
                "section": doc.metadata.get("section"),
                "distance": float(distance)
            }
            for doc, distance in results
        ],
        "retrieval_time": retrieval_time
    }

print("[SUCCESS] Experimental evidence gate initialized!")
```

```python

stress_tests = [
    # Answerable paraphrases
    ("How many academic credits does a CS major need?", True),
    ("Can I graduate in CS without completing both capstone courses?", True),
    ("Which math course is required for the CS major?", True),
    ("What GPA is needed to enroll in Capstone I?", True),
    ("How can I declare a minor in Computer Science?", True),

    # Questions not supported by the indexed handbook text
    ("What is the Wi-Fi password in the CS laboratory?", False),
    ("Who won the university hackathon in October 2026?", False),
    ("What is the salary of the Computer Science program director?", False),
    ("What is the personal email address of the CS department chair?", False),
    ("What is the exact room number of tomorrow's CS311 lecture?", False)
]

stress_records = []

print("=" * 70)
print("ABSTENTION STRESS TEST")
print("=" * 70)

for question, expected_answerable in stress_tests:
    result = check_evidence_sufficiency(
        question,
        threshold=EXPERIMENTAL_DISTANCE_THRESHOLD
    )

    predicted_answerable = result["decision"] == "retrieve"
    correct = predicted_answerable == expected_answerable

    stress_records.append({
        "question": question,
        "expected_answerable": expected_answerable,
        "predicted_answerable": predicted_answerable,
        "top1_distance": result["top1_distance"],
        "correct": correct
    })

    print(f"\nQuestion: {question}")
    print("Expected:", "Answerable" if expected_answerable else "Abstain")
    print("Decision:", result["decision"])
    print(f"Top-1 distance: {result['top1_distance']:.4f}")
    print("Match:", correct)
    print("Top-1 page:", result["sources"][0]["page"]
          if result["sources"] else "None")

stress_df = pd.DataFrame(stress_records)

print("\n" + "=" * 70)
print("STRESS TEST SUMMARY")
print("=" * 70)

print("Total questions:", len(stress_df))
print("Correct decisions:", stress_df["correct"].sum())
print("Decision accuracy:", round(stress_df["correct"].mean(), 4))

print("\nConfusion matrix:")
print(pd.crosstab(
    stress_df["expected_answerable"],
    stress_df["predicted_answerable"],
    rownames=["Expected answerable"],
    colnames=["Predicted answerable"]
))

stress_df.to_csv(
    PROJECT_ROOT / "results" / "abstention_stress_test.csv",
    index=False
)

print("\n[SUCCESS] Stress test completed!")
```

### False Abstention Root-Cause Analysis

A manual inspection of retrieved evidence was conducted for one incorrectly rejected answerable question and two correctly rejected unanswerable questions.

For the capstone question, the Top-1 result from page 8 described the option of taking additional intermediate or advanced courses instead of the capstone. Another retrieved chunk from page 7 explicitly listed the alternative course combinations.

Despite the presence of relevant evidence, the system abstained because the Top-1 distance of 0.6819 exceeded the experimental threshold of 0.68.

In contrast, the retrieved chunks for questions about the laboratory Wi-Fi password and the next day's CS311 classroom did not contain the requested information.

**Discussion:** The findings confirm that a fixed retrieval-distance threshold can reject answerable questions even when relevant evidence is present among the retrieved chunks.

A more cautious decision mechanism should distinguish high-confidence matches, clearly weak matches, and uncertain cases requiring additional verification.

Distance-based classification alone should not be treated as a guarantee that an answer is supported by the source document.

```python

import time

LOW_DISTANCE_BOUND = 0.60
HIGH_DISTANCE_BOUND = 0.75

def three_state_evidence_gate(question, top_k=5):
    start = time.perf_counter()

    results = structured_db.similarity_search_with_score(
        question, k=top_k
    )

    if not results:
        return {
            "question": question,
            "state": "likely_insufficient",
            "top1_distance": None,
            "results": [],
            "retrieval_time": time.perf_counter() - start
        }

    distance = float(results[0][1])

    if distance <= LOW_DISTANCE_BOUND:
        state = "likely_relevant"
    elif distance <= HIGH_DISTANCE_BOUND:
        state = "needs_verification"
    else:
        state = "likely_insufficient"

    return {
        "question": question,
        "state": state,
        "top1_distance": distance,
        "results": results,
        "retrieval_time": time.perf_counter() - start
    }


print("[SUCCESS] Three-state evidence gate initialized!")
```

```python

from collections import Counter

print("=" * 70)
print("THREE-STATE EVIDENCE GATE EVALUATION")
print("=" * 70)

records = []

for question, expected_answerable in stress_tests:
    result = three_state_evidence_gate(question)

    records.append({
        "question": question,
        "expected_answerable": expected_answerable,
        "state": result["state"],
        "distance": result["top1_distance"]
    })

    print(f"\nQuestion: {question}")
    print("Expected answerable:", expected_answerable)
    print("Top-1 distance:", round(result["top1_distance"], 4))
    print("New state:", result["state"])

print("\n" + "=" * 70)
print("STATE DISTRIBUTION")
print("=" * 70)

for (expected, state), count in sorted(
    Counter(
        (item["expected_answerable"], item["state"])
        for item in records
    ).items()
):
    print(
        f"Expected answerable={expected} | "
        f"{state}: {count}"
    )

print("\n[COMPLETE] Three-state diagnostic evaluation finished!")
```

### LLM-Based Evidence Verification: Failure Analysis

An evidence-grounded verification experiment was conducted using Qwen 2.5 0.5B on an answerable question about graduating without completing both Computer Science capstone courses.

The retrieved context included the handbook's explicit capstone alternatives. However, the language model incorrectly answered that the student must complete at least two Computer Science courses, including MATH202 as an eligible course.

This information corresponds to the requirements for formally declaring the Computer Science major, not to the capstone graduation alternatives.

Although the model labeled its response as "SUPPORTED", it failed to provide the requested evidence quotation and produced a factually incorrect answer. Generation required 31.69 seconds on the CPU environment.

**Discussion:** The experiment demonstrates that prompting a small language model to verify its own answer does not guarantee factual grounding. The model may select an unrelated statement from the retrieved context and incorrectly associate it with the question.

A self-reported support label should not be treated as an independently validated result. Additional output validation and direct evidence checking are necessary.

Given the model's latency and reliability limitations, deterministic extraction should remain the preferred method for structured academic facts, while unsupported or ambiguous cases should be handled conservatively.

```python

import re

def validate_verification_output(output, context):
    lines = [line.strip() for line in output.splitlines() if line.strip()]

    if not lines:
        return {
            "valid": False,
            "reason": "Empty verifier output"
        }

    verdict = lines[0].upper()

    if verdict not in {"SUPPORTED", "UNSUPPORTED"}:
        return {
            "valid": False,
            "reason": "Missing valid verdict"
        }

    if verdict == "UNSUPPORTED":
        return {
            "valid": True,
            "verdict": "UNSUPPORTED",
            "reason": "Verifier declined to answer"
        }

    evidence_match = re.search(
        r"(?im)^Evidence:\s*(.+)$",
        output
    )

    answer_match = re.search(
        r"(?ims)^Answer:\s*(.+)$",
        output
    )

    if not evidence_match:
        return {
            "valid": False,
            "reason": "SUPPORTED response missing evidence quotation"
        }

    if not answer_match or not answer_match.group(1).strip():
        return {
            "valid": False,
            "reason": "SUPPORTED response missing answer"
        }

    evidence = evidence_match.group(1).strip().strip('"')

    normalize = lambda s: " ".join(s.split()).casefold()

    if len(evidence) < 15 or normalize(evidence) not in normalize(context):
        return {
            "valid": False,
            "reason": "Evidence quotation not found in retrieved context"
        }

    return {
        "valid": True,
        "verdict": "SUPPORTED",
        "evidence": evidence,
        "answer": answer_match.group(1).strip(),
        "reason": (
            "Output format and quotation verified; "
            "answer entailment NOT verified"
        )
    }


validation = validate_verification_output(
    verification_output,
    verification_context
)

print("=" * 65)
print("VERIFIER OUTPUT VALIDATION")
print("=" * 65)

for key, value in validation.items():
    print(f"{key}: {value}")

assert validation["valid"] is False

print("\n[SUCCESS] Invalid verifier response detected!")
```

### Evidence Verification Output Validation

A deterministic validation function was introduced to check whether the language model's verification response followed the required output format and included an evidence quotation present in the retrieved context.

When applied to the previous capstone verification response, the validator returned `valid: False` because the model labeled its answer as `SUPPORTED` without providing the required evidence quotation.

The validation step successfully identified a structural violation in the verifier output.

However, checking response format and quotation presence is not equivalent to verifying factual correctness or logical entailment. Even a correctly quoted passage may not support the generated answer.

**Decision:** The experimental LLM-based verifier will not be incorporated into the current deployment pipeline. The system will prioritize deterministic extraction for structured academic facts and use conservative evidence handling for general RAG questions.

```python

import time

SAFE_DISTANCE_BOUND = 0.60

ABSTENTION_MESSAGE = (
    "I could not find sufficiently reliable evidence "
    "to answer this question from the provided "
    "Computer Science Major Handbook."
)

def safe_academic_chat(question, top_k=5):
    start = time.perf_counter()

    course_codes = extract_course_codes(question)
    intent = detect_course_intent(question)

    # Route A: Questions about one explicitly named course
    if len(course_codes) == 1:

        code = course_codes[0]

        if intent == "inverse_prerequisite":
            matches = find_courses_requiring(code)

            if matches:
                answer_lines = [
                    (
                        f"{item['course_code']}: "
                        f"{item['course_title']} — "
                        f"Prerequisites: {item['prerequisites']} "
                        f"[PDF pp. {item['page']}-{item['end_page']}]"
                    )
                    for item in matches
                ]

                answer = (
                    f"Courses explicitly listing {code} "
                    "as a prerequisite in the indexed catalogue:\n"
                    + "\n".join(answer_lines)
                )
            else:
                answer = (
                    f"No courses explicitly listing {code} "
                    "as a prerequisite were found in the "
                    "indexed catalogue."
                )

            return {
                "answer": answer,
                "route": "inverse_prerequisite",
                "grounding": "deterministic_extraction",
                "sources": matches,
                "elapsed": time.perf_counter() - start
            }

        course_result = field_aware_course_retrieve(question)

        if course_result["status"] == "success":
            return {
                "answer": course_result["evidence"],
                "route": "course_lookup",
                "grounding": "deterministic_extraction",
                "sources": [course_result["metadata"]],
                "elapsed": time.perf_counter() - start
            }

        return {
            "answer": ABSTENTION_MESSAGE,
            "route": "course_not_found",
            "grounding": "abstained",
            "sources": [],
            "elapsed": time.perf_counter() - start
        }

    # Route B: General academic question
    retrieved = structured_db.similarity_search_with_score(
        question, k=top_k
    )

    if not retrieved:
        return {
            "answer": ABSTENTION_MESSAGE,
            "route": "general_rag",
            "grounding": "abstained",
            "sources": [],
            "elapsed": time.perf_counter() - start
        }

    top1_distance = float(retrieved[0][1])

    # Conservative policy: do not assume that an
    # uncertain retrieval result supports an answer.
    if top1_distance > SAFE_DISTANCE_BOUND:
        return {
            "answer": ABSTENTION_MESSAGE,
            "route": "general_rag",
            "grounding": "abstained",
            "top1_distance": top1_distance,
            "sources": [],
            "elapsed": time.perf_counter() - start
        }

    context = "\n\n---\n\n".join(
        (
            f"[PDF page {doc.metadata['page']}]\n"
            f"{doc.page_content}"
        )
        for doc, distance in retrieved
    )

    prompt = PDF_RAG_PROMPT.format(
        context=context,
        question=question
    )

    generated_answer = pdf_llm.invoke(prompt).strip()

    sources = [
        {
            "source": doc.metadata.get("source"),
            "page": doc.metadata.get("page"),
            "distance": float(distance)
        }
        for doc, distance in retrieved
    ]

    return {
        "answer": generated_answer,
        "route": "general_rag",
        "grounding": "llm_generated_unverified",
        "top1_distance": top1_distance,
        "sources": sources,
        "elapsed": time.perf_counter() - start
    }


print("[SUCCESS] Conservative unified chatbot initialized!")
```

```python

from pathlib import Path

course_lookup_code = r'''
import json
import re
from pathlib import Path


COURSE_CODE_PATTERN = re.compile(
    r"\bCS[\s-]*(\d{3})\b",
    flags=re.IGNORECASE
)

PREREQUISITE_PATTERN = re.compile(
    r"(?im)^[ \t]*Prerequisites?\s*:\s*([^\n]+)"
)


class CourseLookup:
    def __init__(self, jsonl_path):
        self.jsonl_path = Path(jsonl_path)

        if not self.jsonl_path.is_file():
            raise FileNotFoundError(
                f"Course documents not found: {self.jsonl_path}"
            )

        self.courses = {}

        with self.jsonl_path.open(
            "r", encoding="utf-8"
        ) as f:
            for line in f:
                if not line.strip():
                    continue

                record = json.loads(line)
                metadata = record["metadata"]
                code = metadata["course_code"]

                if code in self.courses:
                    raise ValueError(
                        f"Duplicate course entry: {code}"
                    )

                self.courses[code] = record

    @staticmethod
    def extract_course_codes(question):
        matches = COURSE_CODE_PATTERN.findall(question)

        return list(dict.fromkeys(
            f"CS{number}" for number in matches
        ))

    @staticmethod
    def detect_intent(question):
        q = question.lower()

        if re.search(
            r"(which|what)\s+courses?.*"
            r"(require|need|depend on)",
            q
        ):
            return "inverse_prerequisite"

        if "prerequisit" in q or "required before" in q:
            return "prerequisite"

        return "course_overview"

    @staticmethod
    def extract_prerequisite(record):
        match = PREREQUISITE_PATTERN.search(
            record["page_content"]
        )

        return match.group(1).strip() if match else None

    def get_course(self, code):
        return self.courses.get(code.upper())

    def get_prerequisites(self, code):
        record = self.get_course(code)

        if record is None:
            return None

        metadata = record["metadata"]
        prerequisite = self.extract_prerequisite(record)

        # An absent field is not equivalent to "None".
        if prerequisite is None:
            return None

        return {
            "course_code": metadata["course_code"],
            "course_title": metadata["course_title"],
            "prerequisites": prerequisite,
            "page": metadata["page"],
            "end_page": metadata["end_page"],
            "source": metadata["source"]
        }

    def get_description(self, code):
        record = self.get_course(code)

        if record is None:
            return None

        return {
            "content": record["page_content"],
            "metadata": record["metadata"]
        }

    def find_courses_requiring(self, target_code):
        target_code = re.sub(
            r"\s+", "", target_code
        ).upper()

        matches = []

        for record in self.courses.values():
            prerequisite = self.extract_prerequisite(record)

            if prerequisite is None:
                continue

            normalized = re.sub(
                r"\b([A-Za-z]+)[\s-]+(\d{3})\b",
                r"\1\2",
                prerequisite.upper()
            )

            if not re.search(
                rf"(?<![A-Z0-9])"
                rf"{re.escape(target_code)}"
                rf"(?![A-Z0-9])",
                normalized
            ):
                continue

            metadata = record["metadata"]

            matches.append({
                "course_code": metadata["course_code"],
                "course_title": metadata["course_title"],
                "prerequisites": prerequisite,
                "page": metadata["page"],
                "end_page": metadata["end_page"],
                "source": metadata["source"]
            })

        return sorted(
            matches,
            key=lambda item: item["course_code"]
        )
'''

module_path = PROJECT_PATH / "app" / "course_lookup.py"
module_path.write_text(
    course_lookup_code.strip() + "\n",
    encoding="utf-8"
)

print("=" * 65)
print("PYTHON MODULE CREATED")
print("=" * 65)
print("File:", module_path)
print("Lines:", len(module_path.read_text().splitlines()))
print("Size:", round(module_path.stat().st_size / 1024, 2), "KB")
print("\n[SUCCESS] course_lookup.py created!")
```

```python

from pathlib import Path

retrieval_code = r'''
import os
from pathlib import Path

from langchain_chroma import Chroma
from langchain_ollama import OllamaEmbeddings


class HandbookRetriever:
    def __init__(
        self,
        db_path,
        collection_name="cs_handbook_structured",
        embedding_model="nomic-embed-text",
        ollama_base_url=None
    ):
        self.db_path = Path(db_path)
        self.collection_name = collection_name

        self.ollama_base_url = (
            ollama_base_url
            or os.getenv(
                "OLLAMA_BASE_URL",
                "http://127.0.0.1:11434"
            )
        )

        if not self.db_path.is_dir():
            raise FileNotFoundError(
                f"ChromaDB directory not found: {self.db_path}"
            )

        if not (self.db_path / "chroma.sqlite3").is_file():
            raise FileNotFoundError(
                "Existing ChromaDB data not found at "
                f"{self.db_path}. Refusing to create "
                "an empty index."
            )

        self.embeddings = OllamaEmbeddings(
            model=embedding_model,
            base_url=self.ollama_base_url
        )

        self.db = Chroma(
            collection_name=self.collection_name,
            embedding_function=self.embeddings,
            persist_directory=str(self.db_path),
            create_collection_if_not_exists=False
        )

        self.document_count = self.db._collection.count()

        if self.document_count == 0:
            raise RuntimeError(
                "The specified Chroma collection is empty."
            )

    def search(self, question, top_k=5):
        results = self.db.similarity_search_with_score(
            question,
            k=top_k
        )

        return [
            {
                "content": doc.page_content,
                "source": doc.metadata.get("source"),
                "page": doc.metadata.get("page"),
                "section": doc.metadata.get("section"),
                "distance": float(distance)
            }
            for doc, distance in results
        ]

    @staticmethod
    def build_context(results):
        sections = []

        for rank, result in enumerate(results, start=1):
            sections.append(
                f"[Source {rank}: "
                f"{result['source']}, "
                f"PDF page {result['page']}]\n"
                f"{result['content']}"
            )

        return "\n\n---\n\n".join(sections)
'''

retrieval_path = PROJECT_PATH / "app" / "retrieval.py"

retrieval_path.write_text(
    retrieval_code.strip() + "\n",
    encoding="utf-8"
)

print("=" * 65)
print("RETRIEVAL MODULE CREATED")
print("=" * 65)
print("File:", retrieval_path)
print("Lines:", len(retrieval_path.read_text().splitlines()))
print("Size:", round(retrieval_path.stat().st_size / 1024, 2), "KB")
print("\n[SUCCESS] app/retrieval.py created!")
```

```python

from pathlib import Path

chatbot_code = r'''
import os
import time
from pathlib import Path

from langchain_ollama import OllamaLLM

from app.course_lookup import CourseLookup
from app.retrieval import HandbookRetriever


ABSTENTION_MESSAGE = (
    "I could not find sufficiently reliable evidence "
    "to answer this question from the provided "
    "Computer Science Major Handbook."
)


class AcademicChatbot:
    def __init__(
        self,
        project_root,
        ollama_base_url=None,
        llm_model="qwen2.5:0.5b",
        embedding_model="nomic-embed-text",
        distance_threshold=0.60
    ):
        self.project_root = Path(project_root)

        self.ollama_base_url = (
            ollama_base_url
            or os.getenv(
                "OLLAMA_BASE_URL",
                "http://127.0.0.1:11434"
            )
        )

        self.distance_threshold = distance_threshold

        self.courses = CourseLookup(
            self.project_root
            / "data/processed/course_documents.jsonl"
        )

        self.retriever = HandbookRetriever(
            db_path=(
                self.project_root
                / "vector_db/structured"
            ),
            collection_name="cs_handbook_structured",
            embedding_model=embedding_model,
            ollama_base_url=self.ollama_base_url
        )

        self.llm = OllamaLLM(
            model=llm_model,
            base_url=self.ollama_base_url,
            temperature=0,
            num_predict=220
        )

    def _make_result(
        self, answer, route, grounding,
        sources, start, **extra
    ):
        return {
            "answer": answer,
            "route": route,
            "grounding": grounding,
            "sources": sources,
            "elapsed": time.perf_counter() - start,
            **extra
        }

    def ask(self, question, top_k=5):
        start = time.perf_counter()

        question = question.strip()

        if not question:
            return self._make_result(
                "Please enter a question.",
                "invalid_input",
                "not_applicable",
                [],
                start
            )

        course_codes = self.courses.extract_course_codes(
            question
        )
        intent = self.courses.detect_intent(question)

        # Route A: Course-specific questions
        if len(course_codes) == 1:
            code = course_codes[0]
            course_record = self.courses.get_course(code)

            # Only use deterministic lookup when the
            # course exists in the extracted catalogue.
            if course_record is not None:

                if intent == "inverse_prerequisite":
                    matches = (
                        self.courses.find_courses_requiring(
                            code
                        )
                    )

                    if matches:
                        lines = [
                            (
                                f"{item['course_code']}: "
                                f"{item['course_title']} — "
                                f"Prerequisites: "
                                f"{item['prerequisites']} "
                                f"[PDF pp. {item['page']}-"
                                f"{item['end_page']}]"
                            )
                            for item in matches
                        ]

                        answer = (
                            f"Courses explicitly listing "
                            f"{code} as a prerequisite "
                            "in the indexed catalogue:\n"
                            + "\n".join(lines)
                        )
                    else:
                        answer = (
                            f"No direct dependencies on {code} "
                            "were found among the indexed "
                            "course entries."
                        )

                    return self._make_result(
                        answer,
                        "inverse_prerequisite",
                        "deterministic_extraction",
                        matches,
                        start
                    )

                if intent == "prerequisite":
                    item = self.courses.get_prerequisites(
                        code
                    )

                    if item is not None:
                        answer = (
                            f"{item['course_code']}: "
                            f"{item['course_title']}\n"
                            f"Prerequisites: "
                            f"{item['prerequisites']}"
                        )

                        return self._make_result(
                            answer,
                            "course_lookup",
                            "deterministic_extraction",
                            [item],
                            start
                        )

                elif intent == "course_overview":
                    item = self.courses.get_description(
                        code
                    )

                    if item is not None:
                        return self._make_result(
                            item["content"],
                            "course_lookup",
                            "deterministic_extraction",
                            [item["metadata"]],
                            start
                        )

            # If no suitable structured record exists,
            # search other sections of the handbook.
            # For inverse questions, do not substitute
            # semantic similarity for exhaustive lookup.
            if intent == "inverse_prerequisite":
                return self._make_result(
                    ABSTENTION_MESSAGE,
                    "inverse_prerequisite",
                    "abstained",
                    [],
                    start
                )

        # Route B: General semantic RAG
        results = self.retriever.search(
            question, top_k=top_k
        )

        if not results:
            return self._make_result(
                ABSTENTION_MESSAGE,
                "general_rag",
                "abstained",
                [],
                start
            )

        top1_distance = results[0]["distance"]

        if top1_distance > self.distance_threshold:
            return self._make_result(
                ABSTENTION_MESSAGE,
                "general_rag",
                "abstained",
                [],
                start,
                top1_distance=top1_distance
            )

        context = self.retriever.build_context(results)

        prompt = f"""You are an academic handbook assistant.

Answer ONLY using the supplied handbook excerpts.
Do not invent missing facts.

If the excerpts do not contain sufficient information,
say: "The provided handbook does not specify this."

Answer briefly and accurately.
Do not confuse major declaration requirements with
degree completion requirements.

Handbook excerpts:
{context}

Student question:
{question}

Answer:
"""

        answer = self.llm.invoke(prompt).strip()

        sources = [
            {
                "source": item["source"],
                "page": item["page"],
                "section": item["section"],
                "distance": item["distance"]
            }
            for item in results
        ]

        return self._make_result(
            answer,
            "general_rag",
            "llm_generated_unverified",
            sources,
            start,
            top1_distance=top1_distance
        )
'''

chatbot_path = PROJECT_PATH / "app" / "chatbot.py"

chatbot_path.write_text(
    chatbot_code.strip() + "\n",
    encoding="utf-8"
)

print("=" * 65)
print("CHATBOT ENGINE MODULE CREATED")
print("=" * 65)
print("File:", chatbot_path)
print("Lines:", len(chatbot_path.read_text().splitlines()))
print("Size:", round(chatbot_path.stat().st_size / 1024, 2), "KB")
print("\n[SUCCESS] app/chatbot.py created!")
```

```python
from pathlib import Path
from zipfile import ZipFile, ZIP_DEFLATED

PROJECT_PATH = Path(PROJECT_ROOT)

required_files = [
    "Dockerfile",
    "docker-compose.yml",
    ".dockerignore",
    "requirements.txt",
    "app/__init__.py",
    "app/course_lookup.py",
    "app/retrieval.py",
    "app/chatbot.py",
    "app/ui.py",
    "data/processed/course_documents.jsonl",
    "data/processed/structured_chunks.jsonl",
]

for filename in required_files:
    path = PROJECT_PATH / filename
    assert path.is_file(), f"Missing file: {filename}"

db_dir = PROJECT_PATH / "vector_db" / "structured"

assert (db_dir / "chroma.sqlite3").is_file(), (
    "Structured ChromaDB database is missing."
)

package_path = PROJECT_PATH / "CS311_Docker_Deployment.zip"

with ZipFile(package_path, "w", ZIP_DEFLATED) as archive:
    for filename in required_files:
        archive.write(
            PROJECT_PATH / filename,
            arcname=f"CS311_Final_Project/{filename}"
        )

    for path in db_dir.rglob("*"):
        if path.is_file():
            relative = path.relative_to(PROJECT_PATH)
            archive.write(
                path,
                arcname=f"CS311_Final_Project/{relative}"
            )

with ZipFile(package_path) as archive:
    packaged_files = archive.namelist()
    bad_file = archive.testzip()

assert bad_file is None, f"Corrupt ZIP member: {bad_file}"

print("=" * 65)
print("DOCKER DEPLOYMENT PACKAGE")
print("=" * 65)

print("Package:", package_path.name)
print("Size:", round(package_path.stat().st_size / 1024**2, 2), "MB")
print("Files included:", len(packaged_files))

print("\nIncluded application files:")
for name in packaged_files:
    if "vector_db/" not in name:
        print(" ", name)

print("\n[SUCCESS] Deployment package verified!")
```
